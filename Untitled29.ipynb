```python
import os
import gc
import torch
import time
import numpy as np
from tqdm import tqdm
from pathlib import Path
from datasets import Dataset
from sentence_transformers import SentenceTransformer

# === CONFIG ===
TEXT_DIR = Path("/content/text_chunks_final")
CHECKPOINT_DIR = Path("/content/rag_chunks_sbert/per_file")
CHECKPOINT_DIR.mkdir(parents=True, exist_ok=True)

SBERT_MODEL = "all-MiniLM-L6-v2"
BATCH_SIZE = 2048  # Reduce to 1024 if memory is tight
DTYPE = np.float32  # Saves GPU/CPU memory
LOG_PREFIX = "[🔧]"

# === Load SBERT model
device = 'cuda' if torch.cuda.is_available() else 'cpu'
print(f"{LOG_PREFIX} Using device: {device.upper()}")
print(f"{LOG_PREFIX} Loading SBERT model: {SBERT_MODEL} ...")
start_model_load = time.time()
model = SentenceTransformer(SBERT_MODEL, device=device)
model.eval()
print(f"{LOG_PREFIX} ✅ Model loaded in {time.time() - start_model_load:.2f} sec\n")

# === Gather text files
all_files = sorted(TEXT_DIR.glob("chunk_*.txt"))
print(f"{LOG_PREFIX} Found total {len(all_files)} text files.")

# === Detect already processed files (via saved folder names)
already_saved = {p.name for p in CHECKPOINT_DIR.iterdir() if p.is_dir()}
files_to_process = [f for f in all_files if f.stem not in already_saved]
print(f"{LOG_PREFIX} Resuming from checkpoint. {len(files_to_process)} files remaining.\n")

# === Main encoding loop
for i, file_path in enumerate(files_to_process, 1):
    fname = file_path.stem
    print(f"[{i}/{len(files_to_process)}] 🔄 Encoding file: {fname}")

    with open(file_path, "r", encoding="utf-8") as f:
        lines = [line.strip() for line in f if line.strip()]

    total_lines = len(lines)
    if total_lines == 0:
        print(f"   ⚠️ Empty file. Skipping.\n")
        continue

    print(f"   📄 {total_lines} lines found.")
    print(f"   ⚙️  Encoding with batch size {BATCH_SIZE} on {device.upper()} ...")

    start_encode = time.time()
    embeddings = []
    total_batches = (total_lines + BATCH_SIZE - 1) // BATCH_SIZE

    with torch.no_grad():
        for j in tqdm(range(0, total_lines, BATCH_SIZE), desc="   ↪ Batches", mininterval=2.0):
            batch_idx = (j // BATCH_SIZE) + 1
            batch = lines[j:j + BATCH_SIZE]

            batch_embeddings = model.encode(
                batch,
                batch_size=BATCH_SIZE,
                convert_to_numpy=True,
                normalize_embeddings=False,
                show_progress_bar=False,
                device=device
            )

            embeddings.append(batch_embeddings.astype(DTYPE))
            print(f"     🧠 Encoded batch {batch_idx}/{total_batches} ({len(batch)} lines)")

            # Optional GPU/CPU cleanup
            if device == 'cuda':
                torch.cuda.empty_cache()
            gc.collect()

    embeddings = np.vstack(embeddings)
    print(f"   ✅ Encoding complete in {time.time() - start_encode:.2f} sec.")

    # Save embeddings + text as HuggingFace dataset
    start_save = time.time()
    ds = Dataset.from_dict({
        "text": lines,
        "embeddings": embeddings
    })
    file_dir = CHECKPOINT_DIR / fname
    file_dir.mkdir(parents=True, exist_ok=True)
    ds.save_to_disk(str(file_dir))
    print(f"   💾 Saved to: {file_dir} in {time.time() - start_save:.2f} sec.\n")

    # Final cleanup
    del embeddings, ds, lines, batch_embeddings
    gc.collect()
    if device == 'cuda':
        torch.cuda.empty_cache()

print(f"\n🎉✅ ALL DONE. Checkpoints saved to → {CHECKPOINT_DIR}")
```

```python
import os, glob, time, subprocess
from google.colab import files

# 0️⃣ Cleanup any old parts
for f in glob.glob('/content/rag_chunks_sbert.tar.part_*'):
    os.remove(f)

# 1️⃣ Start stream-tar + split (no compression)
cmd = "tar -cf - /content/rag_chunks_sbert | split -b 2G - /content/rag_chunks_sbert.tar.part_"
proc = subprocess.Popen(cmd, shell=True)

downloaded = set()
try:
    while True:
        parts = sorted(glob.glob('/content/rag_chunks_sbert.tar.part_*'))
        for part in parts:
            if part not in downloaded:
                files.download(part)
                os.remove(part)
                downloaded.add(part)
        if proc.poll() is not None:
            break
        time.sleep(1)
    # final sweep
    for part in sorted(glob.glob('/content/rag_chunks_sbert.tar.part_*')):
        if part not in downloaded:
            files.download(part)
            os.remove(part)
            downloaded.add(part)
finally:
    if proc.poll() is None:
        proc.kill()

print(f"✅ Done: {len(downloaded)} parts streamed, downloaded, and removed.")
```
